```python
# Task 2: Library Import
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python
# Task 3: Create the Dataset
data = {
    'Student': ['Ali', 'Ahmed', 'Sara', 'Hina', 'Hamza', 'Ayesha', 'Bilal', 'Maha'],
    'Programming': [78, 65, 88, 74, 55, 92, 70, 81],
    'Statistics': [72, 69, 91, 78, 61, 89, 66, 84],
    'Attendance': [85, 75, 95, 82, 68, 97, 79, 91]
}

df = pd.DataFrame(data)
df
```

```python
# Task 4: Inspect the Data
print("--- First 5 Rows ---")
print(df.head())

print("\n--- DataFrame Shape (Rows, Columns) ---")
print(df.shape)

print("\n--- Column Names & Data Types ---")
print(df.info())

print("\n--- Basic Summary Statistics ---")
print(df.describe())
```

```python
# Task 5: Descriptive Statistics
cols = ['Programming', 'Statistics', 'Attendance']

stats_df = pd.DataFrame({
    'Mean': df[cols].mean(),
    'Median': df[cols].median(),
    'Min': df[cols].min(),
    'Max': df[cols].max(),
    'Range': df[cols].max() - df[cols].min(),
    'Std Dev': df[cols].std()
})

stats_df
```

```python
# Task 6: Visualization
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

# 1. Bar Chart for Average Marks
avg_marks = df[['Programming', 'Statistics']].mean()
sns.barplot(x=avg_marks.index, y=avg_marks.values, ax=axes[0], palette='Blues_d')
axes[0].set_title('Average Marks Comparison')
axes[0].set_ylabel('Average Marks')

# 2. Histogram for Statistics Marks
sns.histplot(df['Statistics'], kde=True, ax=axes[1], color='teal', bins=5)
axes[1].set_title('Distribution of Statistics Marks')
axes[1].set_xlabel('Statistics Marks')

# 3. Boxplot for Programming Marks
sns.boxplot(y=df['Programming'], ax=axes[2], color='salmon')
axes[2].set_title('Spread of Programming Marks')
axes[2].set_ylabel('Programming Marks')

plt.tight_layout()
plt.show()
```

Selected Variable: Programming (or Attendance).

Explanation: Programming is the most suitable variable to examine alongside Statistics because both reflect technical and analytical coursework. Evaluating their relationship helps reveal whether foundational analytical ability transfers across computer science subjects.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set random seed for reproducible fictional data
np.random.seed(42)

# 1. Create a dataset of 20 students
data_20 = {
    'Student_ID': [f'STD_{i:02d}' for i in range(1, 21)],
    'Age': np.random.randint(19, 24, size=20),
    'Attendance_%': np.random.randint(60, 98, size=20),
    'Study_Hours_Per_Week': np.random.randint(5, 25, size=20),
    'Programming_Marks': np.random.randint(50, 96, size=20),
    'Statistics_Marks': np.random.randint(48, 98, size=20),
    'GPA': np.round(np.random.uniform(2.2, 3.9, size=20), 2)
}

df_assignment = pd.DataFrame(data_20)

print("=== FIRST 5 ROWS OF DATASET ===")
print(df_assignment.head())

print("\n=== SUMMARY STATISTICS ===")
print(df_assignment.describe())

# 2. Exploratory Data Analysis (Visualizations)
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Plot 1: Study Hours vs. GPA
sns.scatterplot(
    data=df_assignment,
    x='Study_Hours_Per_Week',
    y='GPA',
    hue='Attendance_%',
    palette='viridis',
    s=100,
    ax=axes[0, 0]
)
axes[0, 0].set_title('Study Hours vs GPA (Colored by Attendance)')

# Plot 2: Distribution of Subject Marks
sns.kdeplot(df_assignment['Programming_Marks'], label='Programming', fill=True, ax=axes[0, 1])
sns.kdeplot(df_assignment['Statistics_Marks'], label='Statistics', fill=True, ax=axes[0, 1])
axes[0, 1].set_title('Density Distribution of Course Marks')
axes[0, 1].legend()

# Plot 3: Boxplot of Key Metrics
sns.boxplot(data=df_assignment[['Programming_Marks', 'Statistics_Marks']], ax=axes[1, 0], palette='Set2')
axes[1, 0].set_title('Spread & Outliers in Subject Marks')

# Plot 4: Correlation Heatmap
sns.heatmap(
    df_assignment[['Age', 'Attendance_%', 'Study_Hours_Per_Week', 'Programming_Marks', 'Statistics_Marks', 'GPA']].corr(),
    annot=True,
    cmap='Blues',
    fmt='.2f',
    ax=axes[1, 1]
)
axes[1, 1].set_title('Correlation Matrix')

plt.tight_layout()
plt.show()
```
